## Step 5 — Look at the results

```python
import json, pandas as pd
from IPython.display import Image, display, Markdown
for task, file in [("Task 1", "Task_1_Bell_State/results/task1_bell_quokka.json"),
                   ("Task 2A", "Task_2_Qiskit_QASM/results/task2a_both_qubits.json"),
                   ("Task 2B", "Task_2_Qiskit_QASM/results/task2b_half_adder.json")]:
    r = json.load(open(file))
    print(f"{task:8} {r['source'][:16]:16} counts = {r['counts']}")
display(Markdown("**Task 3 – recorded games**")); display(pd.read_csv("Task_3_Quantum_TicTacToe/results/task3_games_summary.csv"))
display(Markdown("**Task 4 – quantum vs local**")); display(pd.read_csv("Task_4_Quantum_ML/results/task4_comparison.csv"))
```
